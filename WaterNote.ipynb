```python
import pandas as pd
import matplotlib.pylab as pt
import numpy as np
```

```python
water = pd.read_csv("waterdata.csv")
water.head(10)
```

```python
#x is for rows, y is for the drinkability column only which is the target.
x = water.drop("Drinkability_Score", axis=1)
y = water["Drinkability_Score"]
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

np.random.seed

x = water.drop("Drinkability_Score", axis=1)
y = water["Drinkability_Score"]

xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.2)

model = RandomForestRegressor()
model.fit(xtrain, ytrain)

model.score(xtest, ytest)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
xtrain = scaler.fit_transform(xtrain)
xtest = scaler.transform(xtest)

print(xtrain)
```

```python
# Test 
#Toxic, Average, Perfect :D
sample = pd.DataFrame({
    'pH': [3.5, 8.2, 7.00],
    'Hardness': [450.0, 210.0, 75.0],
    'Solids_TDS': [50000.0, 18000.0, 400.0],
    'Chloramines': [13.0, 7.5, 0.1],
    'Sulfate': [550.0, 280.0, 30.0],
    'Conductivity': [1200.0, 550.0, 80.0],
    'Organic_Carbon': [32.0, 16.0, 0.5],
    'Trihalomethanes': [140.0, 85.0, 1.0],
    'Turbidity': [15.5, 4.2, 0.05]
})

model.predict(sample)
```

```python
import joblib

joblib.dump(model , 'water_qualityM.pkl')
joblib.dump(scaler, 'scaler.pkl')
```
